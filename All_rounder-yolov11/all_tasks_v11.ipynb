## 1. Segementations

### Segmentation is a fundamental task in computer vision that involves dividing an image into its constituent parts or objects. The goal is to simplify the image into meaningful regions, making it easier to analyze and understand.

## Semantic Segmentation
+ #### It assigns a class label to each pixel in the image (e.g., car, tree, person).
+ #### Example: Self-driving cars use semantic segmentation to identify roads, pedestrians, and obstacles

## Instance Segmentation:
+ #### Identifies and separates individual objects of the same class (e.g., multiple cars). 
+ #### Example: Autonomous drones use instance segmentation to detect and track individual objects.

# Note : YOLOv11 performs instance segmentation.

```python
import cv2
import os
os.environ["KMP_DUPLICATE_LIB_OK"]="TRUE"
from ultralytics import YOLO
```

## loading segementation model

```python
model = YOLO("yolo11s-seg.pt")
```

# 2. POSE

### Pose estimation is a task that involves identifying the location of specific points in an image, usually referred to as keypoints. The keypoints can represent various parts of the object such as joints, landmarks, or other distinctive features. The locations of the keypoints are usually represented as a set of 2D [x, y] or 3D [x, y, visible] coordinates.

### The output of a pose estimation model is a set of points that represent the keypoints on an object in the image, usually along with the confidence scores for each point. Pose estimation is a good choice when you need to identify specific parts of an object in a scene, and their location in relation to each other.

```python
from ultralytics import YOLO
pose_model = YOLO("yolo11n-pose.pt")  # load an official model
```

```python
results = pose_model("https://ultralytics.com/images/bus.jpg") 
```

```python
results[0].show()
```

```python
print(results[0].names)
```

```python
print(results[0].boxes)
```

```python
print(results[0].keypoints)
```

# 3. Oriented Detection

### Oriented object detection goes a step further than object detection and introduce an extra angle to locate objects more accurate in an image.

### The output of an oriented object detector is a set of rotated bounding boxes that exactly enclose the objects in the image, along with class labels and confidence scores for each box. Object detection is a good choice when you need to identify objects of interest in a scene, but don't need to know exactly where the object is or its exact shape.

```python
from ultralytics import YOLO
orient_model = YOLO("yolo11n-obb.pt")  # load an official model
```

```python
results = orient_model("https://ultralytics.com/images/boats.jpg") 
```

```python
results[0].show()
```

```python
print(results[0].names)
```

```python
print(results[0].obb)
```

# 4. Classification

### Image classification is a fundamental task in the field of computer vision and artificial intelligence, where the goal is to assign a label or category to an image based on its visual content. This process involves analyzing pixel data in the image using algorithms to identify and categorize objects, scenes, or activities present within it.

```python
from ultralytics import YOLO
classification_model = YOLO("yolo11x-cls.pt")
```

```python
results = classification_model("https://ultralytics.com/images/bus.jpg") 
```

```python
results[0].show()
```

```python
results = classification_model("20190308074624_10-most-expensive-cars.jpg") 
```

```python
results[0].show()
```

```python
print(results[0].names)
```
